## 1.1 Estereoscopia e Percepção de Profundidade

A estereoscopia é uma técnica baseada na captura de duas imagens de uma mesma cena a partir de pontos de vista ligeiramente diferentes. Esse princípio é semelhante ao funcionamento da visão humana, em que cada olho observa o ambiente de uma posição distinta, permitindo ao cérebro interpretar a profundidade dos objetos.

Em sistemas de Visão Computacional, uma câmera estéreo utiliza duas câmeras separadas por uma distância conhecida, chamada baseline. As diferenças entre as imagens capturadas pela câmera esquerda e pela câmera direita permitem estimar a posição tridimensional dos objetos presentes na cena. Assim, a visão estéreo possibilita transformar informações bidimensionais, obtidas pelas imagens, em uma estimativa de profundidade do ambiente observado.

## 1.2 Geometria Epipolar

A geometria epipolar descreve a relação geométrica entre duas imagens obtidas por câmeras posicionadas em pontos distintos. Quando um ponto tridimensional da cena é observado pelas duas câmeras, ele é projetado em uma posição na imagem esquerda e em outra posição correspondente na imagem direita. A geometria epipolar estabelece restrições sobre essa correspondência, permitindo reduzir a região de busca do ponto correspondente.

Um dos principais elementos dessa geometria são os epipolos. Os epipolos são os pontos nos quais a linha que liga os centros ópticos das duas câmeras intercepta os planos de imagem. Em outras palavras, o epipolo da imagem esquerda corresponde à projeção do centro óptico da câmera direita, enquanto o epipolo da imagem direita corresponde à projeção do centro óptico da câmera esquerda. Esses pontos são importantes porque todas as linhas epipolares de uma mesma imagem passam pelo respectivo epipolo.

Outro conceito fundamental é o plano epipolar, definido pelo ponto tridimensional observado e pelos centros ópticos das duas câmeras. Esse plano contém a baseline e os raios de projeção que ligam o ponto real da cena às suas projeções nas imagens esquerda e direita. A interseção desse plano com cada plano de imagem gera as chamadas linhas epipolares.

A linha epipolar representa, portanto, o local onde o ponto correspondente deve estar na outra imagem. Assim, quando um ponto é identificado na imagem esquerda, seu correspondente na imagem direita não precisa ser procurado em toda a imagem, mas apenas sobre a linha epipolar associada. Essa restrição torna o processo de correspondência entre imagens mais eficiente e é uma das bases para o cálculo de profundidade em sistemas estéreo.

## 1.3 Calibração e Disparidade Estéreo

A calibração de câmeras tem como objetivo determinar os parâmetros internos e externos do sistema. Os parâmetros internos descrevem características próprias de cada câmera, como distância focal, centro óptico e distorções da lente. Já os parâmetros externos descrevem a posição e a orientação relativa entre as duas câmeras.

No contexto da geometria epipolar, uma matriz importante é a matriz fundamental, representada por F. Essa matriz possui dimensão 3×3 e relaciona pontos correspondentes entre as duas imagens. Sua função é descrever matematicamente a restrição epipolar, permitindo calcular a linha epipolar em uma imagem a partir de um ponto localizado na outra imagem.

A relação entre dois pontos correspondentes pode ser representada por:

  x2TFx1​ = 0

em que x1 representa um ponto na primeira imagem, x2 representa o ponto correspondente na segunda imagem e F é a matriz fundamental. De forma geral, essa matriz pode ser escrita como:

$$
F =
\begin{bmatrix}
f_{11} & f_{12} & f_{13} \\
f_{21} & f_{22} & f_{23} \\
f_{31} & f_{32} & f_{33}
\end{bmatrix}
$$
	​

Embora apresente nove elementos, a matriz fundamental é definida até um fator de escala e possui posto 2. Seus parâmetros representam a relação projetiva entre as duas câmeras, incorporando informações associadas à orientação relativa, ao deslocamento entre os centros ópticos e aos parâmetros internos das câmeras. Na prática, ela pode ser estimada a partir de pontos correspondentes entre as imagens.

Após a calibração, as imagens podem ser corrigidas e alinhadas por meio da retificação estéreo. Esse processo facilita o cálculo da disparidade, que representa o deslocamento aparente de um mesmo ponto entre a imagem esquerda e a imagem direita. Em um sistema estéreo retificado, esse deslocamento ocorre principalmente na direção horizontal e pode ser expresso por:

  d=xL​ −xR
	​
em que xL é a coordenada horizontal do ponto na imagem esquerda e xR é a coordenada horizontal do ponto correspondente na imagem direita.

A disparidade está diretamente relacionada à profundidade da cena. Objetos mais próximos das câmeras apresentam maior disparidade, enquanto objetos mais distantes apresentam menor disparidade. Dessa forma, ao calcular a disparidade entre as imagens, torna-se possível estimar a distância dos objetos em relação ao sistema estéreo.

## CALIBRAÇÃO DA CÂMERA ESTEROSCÓPICA E GERAÇÃO DA IMAGEM 3D

````python
import numpy as np 
import cv2
from tqdm import tqdm

# Set the path to the images captured by the left and right cameras
pathL = "./data/stereoL/"
pathR = "./data/stereoR/"

print("Extracting image coordinates of respective 3D pattern ....\n")

# Termination criteria for refining the detected corners
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)


objp = np.zeros((9*6,3), np.float32)
objp[:,:2] = np.mgrid[0:9,0:6].T.reshape(-1,2)

img_ptsL = []
img_ptsR = []
obj_pts = []

for i in tqdm(range(1,28)):
	imgL = cv2.imread(pathL+"img%d.png"%i)
	imgR = cv2.imread(pathR+"img%d.png"%i)
	imgL_gray = cv2.imread(pathL+"img%d.png"%i,0)
	imgR_gray = cv2.imread(pathR+"img%d.png"%i,0)

	outputL = imgL.copy()
	outputR = imgR.copy()

	retR, cornersR =  cv2.findChessboardCorners(outputR,(9,6),None)
	retL, cornersL = cv2.findChessboardCorners(outputL,(9,6),None)

	if retR and retL:
		obj_pts.append(objp)
		cv2.cornerSubPix(imgR_gray,cornersR,(11,11),(-1,-1),criteria)
		cv2.cornerSubPix(imgL_gray,cornersL,(11,11),(-1,-1),criteria)
		cv2.drawChessboardCorners(outputR,(9,6),cornersR,retR)
		cv2.drawChessboardCorners(outputL,(9,6),cornersL,retL)
		cv2.imshow('cornersR',outputR)
		cv2.imshow('cornersL',outputL)
		cv2.waitKey(0)

		img_ptsL.append(cornersL)
		img_ptsR.append(cornersR)


print("Calculating left camera parameters ... ")
# Calibrating left camera
retL, mtxL, distL, rvecsL, tvecsL = cv2.calibrateCamera(obj_pts,img_ptsL,imgL_gray.shape[::-1],None,None)
hL,wL= imgL_gray.shape[:2]
new_mtxL, roiL= cv2.getOptimalNewCameraMatrix(mtxL,distL,(wL,hL),1,(wL,hL))

print("Calculating right camera parameters ... ")
# Calibrating right camera
retR, mtxR, distR, rvecsR, tvecsR = cv2.calibrateCamera(obj_pts,img_ptsR,imgR_gray.shape[::-1],None,None)
hR,wR= imgR_gray.shape[:2]
new_mtxR, roiR= cv2.getOptimalNewCameraMatrix(mtxR,distR,(wR,hR),1,(wR,hR))


print("Stereo calibration .....")
flags = 0
flags |= cv2.CALIB_FIX_INTRINSIC
# Here we fix the intrinsic camara matrixes so that only Rot, Trns, Emat and Fmat are calculated.
# Hence intrinsic parameters are the same 

criteria_stereo= (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)


# This step is performed to transformation between the two cameras and calculate Essential and Fundamenatl matrix
retS, new_mtxL, distL, new_mtxR, distR, Rot, Trns, Emat, Fmat = cv2.stereoCalibrate(obj_pts,
                                                          img_ptsL,
                                                          img_ptsR,
                                                          new_mtxL,
                                                          distL,
                                                          new_mtxR,
                                                          distR,
                                                          imgL_gray.shape[::-1],
                                                          criteria_stereo,
                                                          flags)

# Once we know the transformation between the two cameras we can perform stereo rectification
# StereoRectify function
rectify_scale= 1 # if 0 image croped, if 1 image not croped
rect_l, rect_r, proj_mat_l, proj_mat_r, Q, roiL, roiR= cv2.stereoRectify(new_mtxL, distL, new_mtxR, distR,
                                                 imgL_gray.shape[::-1], Rot, Trns,
                                                 rectify_scale,(0,0))

# Use the rotation matrixes for stereo rectification and camera intrinsics for undistorting the image
# Compute the rectification map (mapping between the original image pixels and 
# their transformed values after applying rectification and undistortion) for left and right camera frames
Left_Stereo_Map= cv2.initUndistortRectifyMap(new_mtxL, distL, rect_l, proj_mat_l,
                                             imgL_gray.shape[::-1], cv2.CV_16SC2)
Right_Stereo_Map= cv2.initUndistortRectifyMap(new_mtxR, distR, rect_r, proj_mat_r,
                                              imgR_gray.shape[::-1], cv2.CV_16SC2)


print("Saving paraeters ......")
cv_file = cv2.FileStorage("data/params_py.xml", cv2.FILE_STORAGE_WRITE)
cv_file.write("Left_Stereo_Map_x",Left_Stereo_Map[0])
cv_file.write("Left_Stereo_Map_y",Left_Stereo_Map[1])
cv_file.write("Right_Stereo_Map_x",Right_Stereo_Map[0])
cv_file.write("Right_Stereo_Map_y",Right_Stereo_Map[1])
cv_file.release()
````

````python
import numpy as np 
import cv2

CamL_id = "data/stereoL.mp4"
CamR_id = "data/stereoR.mp4"

CamL = cv2.VideoCapture(CamL_id)
CamR = cv2.VideoCapture(CamR_id)

print("Reading parameters ......")
cv_file = cv2.FileStorage("data/params_py.xml", cv2.FILE_STORAGE_READ)

Left_Stereo_Map_x = cv_file.getNode("Left_Stereo_Map_x").mat()
Left_Stereo_Map_y = cv_file.getNode("Left_Stereo_Map_y").mat()
Right_Stereo_Map_x = cv_file.getNode("Right_Stereo_Map_x").mat()
Right_Stereo_Map_y = cv_file.getNode("Right_Stereo_Map_y").mat()
cv_file.release()

while True:
	retR, imgR = CamR.read()
	retL, imgL = CamL.read()
	
	if retL and retR:
		imgR_gray = cv2.cvtColor(imgR, cv2.COLOR_BGR2GRAY)
		imgL_gray = cv2.cvtColor(imgL, cv2.COLOR_BGR2GRAY)

		Left_nice = cv2.remap(imgL, Left_Stereo_Map_x, Left_Stereo_Map_y, cv2.INTER_LANCZOS4, cv2.BORDER_CONSTANT, 0)
		Right_nice = cv2.remap(imgR, Right_Stereo_Map_x, Right_Stereo_Map_y, cv2.INTER_LANCZOS4, cv2.BORDER_CONSTANT, 0)

		output = Right_nice.copy()
		output[:,:,0] = Right_nice[:,:,0]
		output[:,:,1] = Right_nice[:,:,1]
		output[:,:,2] = Left_nice[:,:,2]

		# output = Left_nice+Right_nice
		output = cv2.resize(output, (700, 700))
		cv2.namedWindow("3D movie", cv2.WINDOW_NORMAL)
		cv2.resizeWindow("3D movie", 700, 700)
		cv2.imshow("3D movie", output)

		# Verifica o ESC enquanto o vídeo roda
		if (cv2.waitKey(1) & 0xFF) == 27:
			break
	
	else:
		# Se os vídeos acabarem, fecha o loop automaticamente
		break

cv2.destroyAllWindows()
````

## Parâmetros Necessários para uma Câmera Estéreo

Com base na teoria de calibração, correção de distorção e nos scripts executados (`calibrate.py` e `movie3d.py`), a operação de um sistema de visão estéreo exige um conjunto de parâmetros matemáticos e ópticos para transformar duas imagens 2D planas em coordenadas 3D reais.

Esses parâmetros são divididos em três grandes blocos:

## 3.2 Parâmetros Intrínsecos (De cada câmera individualmente)

Esses parâmetros dependem exclusivamente da fabricação e da lente de **cada câmera isolada**. Eles não mudam, mesmo se você mexer na posição de uma câmera em relação à outra.

* **Matriz Intrínseca da Câmera ($M$ ou $K$):** É a matriz que faz a ponte entre o mundo real (3D) e os pixels da tela (2D). Ela guarda:
* $f_x, f_y$: A distância focal da lente, medida em pixels para os eixos horizontal e vertical.
* $c_x, c_y$: O ponto principal, que representa o centro óptico do sensor da câmera (onde o raio de luz central bate).


* **Coeficientes de Distorção da Lente:** Como as lentes físicas não são perfeitas, elas distorcem a imagem. A calibração calcula coeficientes para corrigir isso:
* **Distarção Radial ($k_1, k_2, k_3$):** Deformação que entorta linhas retas nas bordas da imagem (causando o efeito de "olho de peixe" ou "barril").
* **Distorção Tangencial ($p_1, p_2$):** Acontece se a lente e o sensor interno da câmera não estiverem perfeitamente paralelos durante a montagem na fábrica.



---

## 3.3 Parâmetros Extrínsecos (Relação entre o par de câmeras)

Esses parâmetros descrevem a **posição física** e a **orientação espacial** da câmera direita em relação à câmera esquerda.

* **Matriz de Rotação ($R$ ou $Rot$):** Mede o ângulo de inclinação entre as duas câmeras (se uma está levemente apontada mais para cima, para o lado ou girada em relação à outra). Serve para alinhar os eixos ópticos.
* **Vetor de Translação ($T$ ou $Trns$):** Mede a distância física em linha reta entre os centros das duas câmeras nos eixos X, Y e Z. A distância no eixo X é chamada de **Baseline** (linha de base) e é o parâmetro mais importante para o cálculo de profundidade (quanto maior a distância entre as câmeras, melhor a precisão em longas distâncias).

---

## 3.3 Matrizes de Retificação e Reprojeção 3D

São os parâmetros gerados pelas funções `cv2.stereoCalibrate` e `cv2.stereoRectify` para alinhar os vídeos em tempo real.

* **Matrizes de Retificação ($R_1, R_2$) e Projeção ($P_1, P_2$):** Alinham matematicamente as imagens das duas câmeras de modo que as linhas de pixels fiquem perfeitamente horizontais e paralelas. Isso faz com que um objeto visto na linha 50 da câmera esquerda esteja exatamente na linha 50 da câmera direita, facilitando a busca por correspondência.
* **Matriz de Reprojeção Disparidade-para-Profundidade ($Q$):** É uma matriz $4 \times 4$ crucial. Ela pega a "disparidade" (a diferença de pixels de um mesmo objeto entre as duas imagens) e calcula a distância real em metros ou centímetros (coordenadas $X, Y, Z$).

````python
import numpy as np 
import cv2
from tqdm import tqdm

# Caminho das imagens salvas pelo script anterior
pathL = "./data/stereoL/"
pathR = "./data/stereoR/"

print("Extracting image coordinates of respective 3D pattern ....\n")

# Critério de parada para o refinamento dos cantos
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)

# IMPORTANTE: Ajustado para o padrão (8,6) do seu primeiro script.
# Se o seu tabuleiro for diferente, mude aqui e no findChessboardCorners.
board_size = (8, 6)

objp = np.zeros((board_size[0] * board_size[1], 3), np.float32)
objp[:, :2] = np.mgrid[0:board_size[0], 0:board_size[1]].T.reshape(-1, 2)

img_ptsL = []
img_ptsR = []
obj_pts = []

# Loop ajustado de 0 a 14 (para as 15 fotos tiradas)
# Se você tirou uma quantidade diferente (ex: 12), mude o 15 para o total de fotos.
total_fotos = 15

for i in tqdm(range(total_fotos)):
    # Lendo os arquivos com a nomenclatura exata do primeiro script
    filename = f"frm_{i}_Duda.jpg"
    
    imgL = cv2.imread(pathL + filename)
    imgR = cv2.imread(pathR + filename)
    
    # Validação caso alguma foto esteja faltando ou com problemas
    if imgL is None or imgR is None:
        print(f"\n[Aviso] Arquivo {filename} não encontrado. Pulando...")
        continue

    imgL_gray = cv2.cvtColor(imgL, cv2.COLOR_BGR2GRAY)
    imgR_gray = cv2.cvtColor(imgR, cv2.COLOR_BGR2GRAY)

    outputL = imgL.copy()
    outputR = imgR.copy()

    # Encontrando os cantos do tabuleiro (ajustado para board_size)
    retR, cornersR = cv2.findChessboardCorners(outputR, board_size, None)
    retL, cornersL = cv2.findChessboardCorners(outputL, board_size, None)

    if retR and retL:
        obj_pts.append(objp)
        cv2.cornerSubPix(imgR_gray, cornersR, (11, 11), (-1, -1), criteria)
        cv2.cornerSubPix(imgL_gray, cornersL, (11, 11), (-1, -1), criteria)
        
        cv2.drawChessboardCorners(outputR, board_size, cornersR, retR)
        cv2.drawChessboardCorners(outputL, board_size, cornersL, retL)
        
        cv2.imshow('cornersR', outputR)
        cv2.imshow('cornersL', outputL)
        
        # Pressione qualquer tecla para ir para a próxima foto do loop
        cv2.waitKey(100)  # Mudei para 100ms para passar mais rápido, se quiser manual mude para 0

        img_ptsL.append(cornersL)
        img_ptsR.append(cornersR)

cv2.destroyAllWindows()

# Garante que pelo menos algumas fotos foram detectadas antes de calibrar
if len(obj_pts) == 0:
    print("Erro: Nenhum padrão de xadrez foi encontrado nas imagens. Verifique a iluminação ou o tamanho do grid.")
    exit(-1)

print(f"\nImagens processadas com sucesso para calibração: {len(obj_pts)}/{total_fotos}")

print("Calculating left camera parameters ... ")
# Calibrating left camera
retL, mtxL, distL, rvecsL, tvecsL = cv2.calibrateCamera(obj_pts, img_ptsL, imgL_gray.shape[::-1], None, None)
hL, wL = imgL_gray.shape[:2]
new_mtxL, roiL = cv2.getOptimalNewCameraMatrix(mtxL, distL, (wL, hL), 1, (wL, hL))

print("Calculating right camera parameters ... ")
# Calibrating right camera
retR, mtxR, distR, rvecsR, tvecsR = cv2.calibrateCamera(obj_pts, img_ptsR, imgR_gray.shape[::-1], None, None)
hR, wR = imgR_gray.shape[:2]
new_mtxR, roiR = cv2.getOptimalNewCameraMatrix(mtxR, distR, (wR, hR), 1, (wR, hR))

print("Stereo calibration .....")
flags = 0
flags |= cv2.CALIB_FIX_INTRINSIC

criteria_stereo = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)

# Calibração estéreo
retS, new_mtxL, distL, new_mtxR, distR, Rot, Trns, Emat, Fmat = cv2.stereoCalibrate(
    obj_pts, img_ptsL, img_ptsR, new_mtxL, distL, new_mtxR, distR,
    imgL_gray.shape[::-1], criteria_stereo, flags
)

# Retificação Estéreo
rectify_scale = 1 
rect_l, rect_r, proj_mat_l, proj_mat_r, Q, roiL, roiR = cv2.stereoRectify(
    new_mtxL, distL, new_mtxR, distR, imgL_gray.shape[::-1], Rot, Trns, rectify_scale, (0, 0)
)

# Mapeamento de Retificação e Remoção de Distorção
Left_Stereo_Map = cv2.initUndistortRectifyMap(new_mtxL, distL, rect_l, proj_mat_l, imgL_gray.shape[::-1], cv2.CV_16SC2)
Right_Stereo_Map = cv2.initUndistortRectifyMap(new_mtxR, distR, rect_r, proj_mat_r, imgR_gray.shape[::-1], cv2.CV_16SC2)

print("Saving parameters ......")
cv_file = cv2.FileStorage("data/params_py.xml", cv2.FILE_STORAGE_WRITE)
cv_file.write("Left_Stereo_Map_x", Left_Stereo_Map[0])
cv_file.write("Left_Stereo_Map_y", Left_Stereo_Map[1])
cv_file.write("Right_Stereo_Map_x", Right_Stereo_Map[0])
cv_file.write("Right_Stereo_Map_y", Right_Stereo_Map[1])
cv_file.release()

print("Parâmetros salvos com sucesso em data/params_py.xml!")
````

# 3.5 Inversão Tridimensional (Pseudoscopia)

## O Problema Identificado

Durante os testes iniciais com os óculos anáglifos (Lente Vermelha no olho esquerdo e Lente Azul/Ciano no olho direito), a equipe detectou uma **inversão na percepção de profundidade**. Objetos que deveriam parecer mais próximos da câmera davam a ilusão de estar no fundo, enquanto o plano de fundo parecia saltar para a frente.

Esse fenômeno óptico é conhecido na literatura de visão computacional como **Pseudoscopia** (ou inversão de relevo).

---

## A Causa do Erro (Mapeamento Incorreto)

O OpenCV trabalha nativamente com as matrizes de imagem no padrão de cores **BGR** (canal `0` = *Blue*/Azul, canal `1` = *Green*/Verde, canal `2` = *Red*/Vermelho).

No primeiro algoritmo implementado, a fusão dos canais foi programada da seguinte forma:

```python
output[:,:,0] = imgR[:,:,0] # Canal Azul recebendo a Câmera Direita
output[:,:,1] = imgR[:,:,1] # Canal Verde recebendo a Câmera Direita
output[:,:,2] = imgL[:,:,2] # Canal Vermelho recebendo a Câmera Esquerda

```

Embora pareça intuitivo associar o canal Vermelho à câmera Esquerda (`imgL`), a física dos filtros dos óculos na tela gera o efeito oposto:

* **A Lente Vermelha (Olho Esquerdo):** Absorve a luz azul e verde, permitindo que o olho enxergue apenas as variações do canal **Vermelho**.
* **A Lente Azul/Ciano (Olho Direito):** Absorve a luz vermelha, permitindo que o olho enxergue apenas as variações dos canais **Azul e Verde**.

No código antigo, o canal Vermelho continha a perspectiva da esquerda, mas quem o lia era o olho **direito** através da lente azul (enxergando-o como sombra escura). Isso fez com que o cérebro recebesse as perspectivas dos olhos trocadas, gerando a inversão do paralaxe.

---

## A Solução Aplicada (Reversão de Canais)

Para corrigir o paralaxe e alinhar o sistema computacional à anatomia visual humana com os óculos, os canais de captura foram invertidos no loop de processamento:

```python
output[:,:,0] = imgL[:,:,0] # Canal Azul passa a receber a Câmera Esquerda
output[:,:,1] = imgL[:,:,1] # Canal Verde passa a receber a Câmera Esquerda
output[:,:,2] = imgR[:,:,2] # Canal Vermelho passa a receber a Câmera Direita

```

### Mecanismo de Filtragem Resultante:

1. **Olho Esquerdo (Filtro Vermelho):** Passa a decodificar na tela o canal Vermelho que agora transporta a imagem da **Câmera Direita**.
2. **Olho Direito (Filtro Ciano):** Passa a decodificar os canais Azul e Verde que agora transportam a imagem da **Câmera Esquerda**.

Essa inversão matemática compensa a filtragem física dos óculos. Como resultado, o córtex visual passou a receber a disparidade binocular correta, restabelecendo a percepção real de relevo, volume e profundidade no vídeo gravado em `.mp4` e `.avi`.

````python
import numpy as np 
import cv2
import os
import time

# IDs das webcams físicas (0 e 2 conforme os testes)
CamL_id = 0
CamR_id = 2

CamL = cv2.VideoCapture(CamL_id)
CamR = cv2.VideoCapture(CamR_id)

# Captura um frame inicial para descobrir a resolução real das câmeras
retL, imgL_init = CamL.read()
if not retL:
    print("Erro crítico: Não foi possível acessar as câmeras. Verifique os cabos USB.")
    exit(-1)

h_live, w_live = imgL_init.shape[:2]
print(f"Câmeras prontas. Resolução de gravação: {w_live}x{h_live}")

# Garante que a pasta data existe
if not os.path.exists('data'):
    os.makedirs('data')

# --- CONFIGURAÇÃO DOS GRAVADORES DUPLOS ---
# 1. Gravador MP4
fourcc_mp4 = cv2.VideoWriter_fourcc(*'mp4v')
filename_mp4 = 'data/video_3d_analifo_garrafa.mp4'
writer_mp4 = cv2.VideoWriter(filename_mp4, fourcc_mp4, 20.0, (w_live, h_live))

# 2. Gravador AVI
fourcc_avi = cv2.VideoWriter_fourcc(*'XVID')
filename_avi = 'data/video_3d_analifo_garrafa.avi'
writer_avi = cv2.VideoWriter(filename_avi, fourcc_avi, 20.0, (w_live, h_live))

if not writer_mp4.isOpened() or not writer_avi.isOpened():
    print("Erro: O OpenCV não conseguiu inicializar um dos gravadores de vídeo.")
    exit(-1)

print("\n--- INICIANDO GRAVAÇÃO DUPLA REVERTIDA/CORRIGIDA ---")
print("-> O efeito 3D agora está alinhado corretamente com os óculos.")
print(f"-> Gravando MP4 em: {filename_mp4}")
print(f"-> Gravando AVI em: {filename_avi}")
print("A gravação vai durar exatamente 15 segundos...")

start_time = time.time()
duracao_segundos = 15 

while True:
    retR, imgR = CamR.read()
    retL, imgL = CamL.read()
    
    if not retL or not retR:
        print("Erro de sinal no meio da gravação.")
        break

    # Calcula o cronômetro
    tempo_passado = time.time() - start_time
    tempo_restante = int(duracao_segundos - tempo_passado)

    # Condição de parada automática (15 segundos)
    if tempo_passado >= duracao_segundos:
        print("\nTempo limite de 15 segundos atingido!")
        break

    # === ALGORITMO DE MONTAGEM DO ANÁGLIFO CORRIGIDO ===
    output = np.zeros_like(imgR)
    
    # Inversão correta para óculos Vermelho/Azul padrão no OpenCV (BGR):
    output[:,:,0] = imgL[:,:,0] # Canal Azul (Blue) vem da câmera ESQUERDA
    output[:,:,1] = imgL[:,:,1] # Canal Verde (Green) vem da câmera ESQUERDA
    output[:,:,2] = imgR[:,:,2] # Canal Vermelho (Red) vem da câmera DIREITA

    # Grava o frame corrigido nos dois arquivos de forma simultânea
    writer_mp4.write(output)
    writer_avi.write(output)
    
    # Cria uma cópia limpa para a janela de exibição com o texto do cronômetro
    tela_exibicao = output.copy()
    cv2.putText(tela_exibicao, f"Gravando 3D Correto: {tempo_restante}s restantes", (30, 50), 
                cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)
    
    # Mostra o resultado final na tela
    cv2.imshow("Gravador Duplo 3D - Canais Corrigidos", tela_exibicao)

    # Permite interromper a gravação antes se pressionar 'q'
    if (cv2.waitKey(1) & 0xFF) == ord('q'):
        print("\nGravação interrompida manualmente.")
        break

# Desliga as câmeras e fecha os arquivos de vídeo com segurança
CamR.release()
CamL.release()
writer_mp4.release()
writer_avi.release()
cv2.destroyAllWindows()

print("\n--- PROCESSAMENTO FINALIZADO ---")
print(f"1. Arquivo salvo: {filename_avi}")
print(f"2. Arquivo salvo: {filename_mp4}")
````

No vídeo gravado, o efeito anáglifo ainda pôde ser observado, porém com uma leve perda de nitidez e definição.

Essa diferença pode estar relacionada à compressão do arquivo `.mp4`, à taxa de quadros da gravação, às limitações das webcams utilizadas e a pequenos desalinhamentos entre as câmeras. Como a montagem foi feita manualmente em uma caixa de sapato, presa com fita adesiva, pequenas diferenças de inclinação ou posição entre as webcams podem ter influenciado o alinhamento das imagens esquerda e direita.

De modo geral, o experimento demonstrou que a câmera estéreo construída foi capaz de gerar um vídeo 3D anáglifo funcional. Foi possível observar o princípio da visão estéreo na prática, relacionando a diferença entre as imagens esquerda e direita com a percepção de profundidade.

# ANÁLISE E DISCUSSÃO DOS ESTUDOS REALIZADOS

A realização do experimento permitiu relacionar os conceitos teóricos de estereoscopia, geometria epipolar, calibração de câmeras e disparidade estéreo com uma aplicação prática. A partir da montagem de uma câmera estéreo, foi possível observar como duas imagens capturadas de pontos de vista diferentes podem ser combinadas para produzir uma percepção tridimensional.

Durante a etapa de construção da câmera estéreo, verificou-se que a estabilidade mecânica da estrutura é um fator fundamental para o sucesso do experimento. Como as webcams foram fixadas manualmente com fita adesiva, pequenos desalinhamentos entre as câmeras podem ter influenciado os resultados obtidos. Em sistemas estéreo, qualquer alteração na posição relativa entre as câmeras após a calibração modifica os parâmetros estimados, prejudicando a retificação das imagens e a percepção de profundidade.

A calibração foi realizada a partir de imagens de um padrão conhecido, no caso, o tabuleiro de calibração. Esse procedimento permitiu estimar os parâmetros intrínsecos de cada câmera, como distância focal, centro óptico e coeficientes de distorção, além dos parâmetros extrínsecos que descrevem a relação espacial entre as duas webcams. A obtenção das matrizes de rotação, translação, matriz essencial e matriz fundamental confirmou a relação geométrica existente entre as duas imagens capturadas.

Um ponto importante observado foi que nem todas as imagens capturadas foram utilizadas com sucesso na calibração. Em um dos pares de imagens, o padrão do tabuleiro não foi reconhecido corretamente em uma das câmeras, sendo necessário descartar esse par. Isso demonstra que a qualidade das imagens de calibração interfere diretamente no resultado final. Fatores como iluminação, foco, posicionamento do tabuleiro, inclinação e nitidez afetam a detecção dos pontos de referência pelo algoritmo.

A matriz fundamental obtida representa a relação epipolar entre as imagens esquerda e direita. Na prática, essa matriz permite restringir a busca por pontos correspondentes a uma linha epipolar, reduzindo a complexidade do problema de correspondência. Esse conceito foi essencial para compreender como a visão estéreo utiliza a diferença entre as duas imagens para estimar profundidade.

Após a calibração, foram gerados os mapas de retificação e correção de distorção, posteriormente salvos no arquivo XML. Esses mapas foram utilizados para alinhar as imagens das duas câmeras e corrigir deformações causadas pelas lentes. A retificação é uma etapa importante, pois facilita a comparação entre as imagens esquerda e direita, fazendo com que os pontos correspondentes fiquem aproximadamente na mesma linha horizontal.

Na etapa de geração do vídeo 3D anáglifo, observou-se que o efeito tridimensional foi perceptível, principalmente em cenas com objetos posicionados em diferentes planos de profundidade. Objetos localizados a uma distância intermediária das câmeras apresentaram melhor percepção de profundidade, enquanto objetos muito próximos causaram maior separação entre os canais vermelho e ciano, dificultando a fusão visual. Já objetos muito distantes apresentaram pouca disparidade, tornando o efeito 3D menos evidente.

Comparando a visualização ao vivo com o vídeo gravado, percebeu-se que o resultado ao vivo apresentou uma percepção de profundidade mais clara e imediata. No vídeo gravado, o efeito anáglifo foi preservado, porém com uma leve perda de nitidez e definição. Essa diferença pode estar associada à compressão do arquivo em formato `.mp4`, à taxa de quadros da gravação, à qualidade das webcams e ao alinhamento físico das câmeras.

De modo geral, os resultados demonstram que a câmera estéreo construída foi funcional e permitiu observar, de forma prática, os princípios fundamentais da visão estéreo. Apesar das limitações da montagem artesanal e da calibração, o experimento foi suficiente para evidenciar a relação entre disparidade, geometria epipolar e percepção de profundidade.


# CONCLUSÕES

O experimento realizado permitiu compreender, na prática, os principais conceitos envolvidos em sistemas de câmera estéreo. A partir da construção de uma estrutura simples com duas webcams USB, foi possível desenvolver uma câmera estereoscópica didática e aplicar técnicas de calibração, retificação e geração de imagem 3D anáglifa.

A revisão teórica sobre geometria epipolar foi essencial para entender a relação entre as imagens capturadas pelas duas câmeras. Conceitos como epipolos, plano epipolar, linhas epipolares, matriz fundamental e disparidade estéreo mostraram-se fundamentais para explicar como um sistema computacional pode estimar profundidade a partir de duas imagens bidimensionais.

A etapa de calibração permitiu obter os parâmetros internos de cada câmera e os parâmetros externos do sistema estéreo. Esses valores foram importantes para corrigir distorções, alinhar as imagens e gerar os mapas de retificação utilizados posteriormente na visualização tridimensional. A calibração também evidenciou a importância de capturar imagens de boa qualidade, com iluminação adequada e correta visualização do padrão de calibração.

A geração do vídeo 3D anáglifo demonstrou que é possível obter percepção de profundidade mesmo com uma montagem simples e de baixo custo. No entanto, o resultado também mostrou limitações relacionadas ao alinhamento manual das câmeras, à qualidade das webcams, à estabilidade da estrutura e à compressão do vídeo gravado. Esses fatores podem reduzir a nitidez da imagem e dificultar a fusão dos canais vermelho e ciano pelos óculos anáglifos.

Conclui-se que o experimento atingiu seus principais objetivos, pois permitiu estudar os fundamentos da estereoscopia, construir uma câmera estéreo simples, realizar sua calibração e gerar uma representação visual tridimensional. Embora o sistema apresente limitações, ele se mostrou adequado para fins didáticos e para a compreensão dos princípios básicos de visão estéreo e reconstrução de profundidade em Visão Computacional.
