# 1. INTRODUÇÃO

Este relatório documenta o experimento de **calibração de câmeras**, cujo objetivo é estimar os parâmetros geométricos que descrevem como uma câmera projeta o mundo tridimensional em uma imagem bidimensional. Esses parâmetros dividem-se em duas categorias:

- **Parâmetros intrínsecos**: dependem exclusivamente da câmera (distância focal, ponto principal, *skew* e coeficientes de distorção de lente). São representados pela matriz **K** (matriz da câmera) e pelo vetor de coeficientes de distorção **dist**.
- **Parâmetros extrínsecos**: descrevem a posição e orientação da câmera em relação a um sistema de coordenadas do mundo (referência), representados pela matriz de rotação **R** e pelo vetor de translação **t**.

A relação entre um ponto 3D do mundo $X_w$ e sua projeção em pixel $x_{img}$ é dada, no modelo *pinhole*, por:

$$ x_{img} \sim K \, [R \mid t] \, X_w $$

Como lentes reais não são ideais, a imagem sofre **distorção radial** (a lente curva mais a luz nas bordas do que no centro, produzindo o efeito de "barril" ou "coxim") e **distorção tangencial** (pequeno desalinhamento entre a lente e o sensor). Essas distorções são modeladas pelo vetor **dist** = $(k_1, k_2, p_1, p_2, k_3)$, sendo $k_i$ os coeficientes radiais e $p_i$ os tangenciais.

O método clássico de calibração (Zhang, 2000), implementado pela função `cv2.calibrateCamera()` do OpenCV, utiliza múltiplas fotografias de um padrão plano de geometria conhecida — no nosso caso, um tabuleiro de xadrez impresso (arquivo `pattern.pdf`) — fotografado em diversas poses. Como as coordenadas 3D dos cantos do tabuleiro são conhecidas (a menos de escala) e suas projeções 2D são detectadas automaticamente pela função `cv2.findChessboardCorners()`, o algoritmo resolve um sistema de otimização não linear que retorna simultaneamente **K**, **dist** e um par (**R**, **t**) para cada imagem utilizada.

Nas seções seguintes, replicamos esse processo em quatro etapas, conforme o roteiro do Laboratório 4: **(A)** revisão do funcionamento do script de calibração fornecido; **(B)** calibração da webcam do grupo com imagens próprias; **(C)** calibração de uma segunda câmera pessoal (smartphone); e **(D)** aplicação dos parâmetros obtidos para corrigir a distorção radial de fotografias coloridas.

## 2.1 Item (A) — Calibração com as imagens de exemplo fornecidas

O script `L4_cal.py`, fornecido pelo professor, implementa o fluxo padrão de calibração do OpenCV descrito na referência [2]. Seu funcionamento é o seguinte:

1. Define a geometria do tabuleiro (`CHECKERBOARD = (6,9)`, o valor original do script) e cria o vetor `objp` com as coordenadas 3D dos cantos do tabuleiro no seu próprio sistema de referência (Z = 0, já que o tabuleiro é plano), com unidade arbitrária igual ao lado de um quadrado.
2. Percorre todas as imagens `*.jpg` da pasta corrente, converte cada uma para escala de cinza e chama `cv2.findChessboardCorners()` para localizar os cantos internos do tabuleiro.
3. Quando os cantos são encontrados (`ret == True`), refina a posição de cada canto com sub-pixel (`cv2.cornerSubPix()`) e armazena o par de correspondências: `objpoints` (3D, iguais para todas as imagens) e `imgpoints` (2D, projeção detectada em cada imagem).
4. Ao final do laço, chama `cv2.calibrateCamera(objpoints, imgpoints, tamanho_da_imagem, None, None)`, que resolve o problema de otimização e devolve:
   - `ret`: o erro de reprojeção RMS (*root-mean-square*) do ajuste;
   - `mtx` → matriz **K** (3×3, parâmetros intrínsecos);
   - `dist` → vetor **dist** (coeficientes de distorção, formato 1×5: $k_1,k_2,p_1,p_2,k_3$);
   - `rvecs` → lista de vetores de rotação (um por imagem, em notação de Rodrigues, 3×1);
   - `tvecs` → lista de vetores de translação (um por imagem, 3×1).

**Significado de cada saída:**

| Parâmetro | Formato | Significado |
|---|---|---|
| **K** | matriz 3×3 | Matriz intrínseca. Contém $f_x, f_y$ (distância focal em pixels nos eixos x/y), $c_x, c_y$ (coordenadas do ponto principal, projeção do centro óptico no sensor) e o termo de *skew* (não-ortogonalidade dos eixos do sensor). |
| **R** | matriz 3×3 (ou vetor de Rodrigues 3×1) | Matriz de rotação extrínseca: descreve a orientação do sistema de coordenadas do tabuleiro (mundo) em relação ao sistema de coordenadas da câmera, para aquela imagem específica. |
| **t** | vetor 3×1 | Vetor de translação extrínseco: descreve a posição da origem do tabuleiro em relação ao centro óptico da câmera, para aquela imagem específica. |
| **dist** | vetor 1×5 | Coeficientes de distorção radial ($k_1,k_2,k_3$) e tangencial ($p_1,p_2$) da lente. |

As imagens de exemplo fornecidas pelo professor (conjunto clássico `left01.jpg` … `left14.jpg`, 13 fotografias válidas de um tabuleiro 9×6 casas / **6×9 cantos internos**, resolução 640×480 px) foram utilizadas a seguir, executando o `L4_cal.py` sem alterações (sem as chamadas de `cv2.imshow`/`cv2.waitKey`, que exigem interface gráfica interativa e foram substituídas por `matplotlib` para exibição inline).

````python
import cv2
import numpy as np
import glob
import pandas as pd

CHECKERBOARD = (6, 9)  # valor original do L4_cal.py, sem alteração
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)

objpoints_A = []
imgpoints_A = []

objp_A = np.zeros((1, CHECKERBOARD[0] * CHECKERBOARD[1], 3), np.float32)
objp_A[0, :, :2] = np.mgrid[0:CHECKERBOARD[0], 0:CHECKERBOARD[1]].T.reshape(-1, 2)

images_A = sorted(glob.glob('sample_professor/*.jpg'))
print(f"Total de imagens de exemplo encontradas: {len(images_A)}")

gray_A = None
h_A = w_A = 0
n_ok_A = 0
for fname in images_A:
    img = cv2.imread(fname)
    h_A, w_A = img.shape[:2]
    gray_A = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    ret, corners = cv2.findChessboardCorners(
        gray_A, CHECKERBOARD,
        cv2.CALIB_CB_ADAPTIVE_THRESH + cv2.CALIB_CB_FAST_CHECK + cv2.CALIB_CB_NORMALIZE_IMAGE
    )
    if ret:
        n_ok_A += 1
        objpoints_A.append(objp_A)
        corners2 = cv2.cornerSubPix(gray_A, corners, (11, 11), (-1, -1), criteria)
        imgpoints_A.append(corners2)

print(f"Cantos detectados com sucesso em {n_ok_A}/{len(images_A)} imagens.")
print(f"Resolução das imagens: {w_A} x {h_A} px")
````

````python
import matplotlib.pyplot as plt

# Exemplo visual: cantos detectados em uma das imagens de exemplo
fname_exemplo_A = images_A[0]
img_exemplo_A = cv2.imread(fname_exemplo_A)
gray_exemplo_A = cv2.cvtColor(img_exemplo_A, cv2.COLOR_BGR2GRAY)
ret_exA, corners_exA = cv2.findChessboardCorners(gray_exemplo_A, CHECKERBOARD,
    cv2.CALIB_CB_ADAPTIVE_THRESH + cv2.CALIB_CB_FAST_CHECK + cv2.CALIB_CB_NORMALIZE_IMAGE)
corners_exA2 = cv2.cornerSubPix(gray_exemplo_A, corners_exA, (11,11), (-1,-1), criteria)
img_corners_A = cv2.drawChessboardCorners(img_exemplo_A.copy(), CHECKERBOARD, corners_exA2, ret_exA)

plt.figure(figsize=(7,5))
plt.imshow(cv2.cvtColor(img_corners_A, cv2.COLOR_BGR2RGB))
plt.title(f"Cantos detectados — {fname_exemplo_A} (exemplo do professor)")
plt.axis('off')
plt.show()
````

````python
ret_A, mtx_A, dist_A, rvecs_A, tvecs_A = cv2.calibrateCamera(
    objpoints_A, imgpoints_A, gray_A.shape[::-1], None, None
)

print("Erro de reprojeção RMS (ret):", ret_A)
print()
print("Matriz K (intrínseca):\n", mtx_A)
print()
print("Vetor dist [k1, k2, p1, p2, k3]:\n", dist_A)
````

````python
# Vetores extrínsecos R e t — um par por imagem utilizada
resumo_A = []
for idx, (rv, tv) in enumerate(zip(rvecs_A, tvecs_A)):
    R_mat, _ = cv2.Rodrigues(rv)
    resumo_A.append({
        "imagem": images_A[idx],
        "rvec (Rodrigues)": np.round(rv.ravel(), 3).tolist(),
        "tvec": np.round(tv.ravel(), 3).tolist(),
    })

df_extrinsecos_A = pd.DataFrame(resumo_A)
df_extrinsecos_A
````

````python
# Exemplo: matriz de rotação R completa (3x3) da primeira imagem, obtida via Rodrigues
R0_A, _ = cv2.Rodrigues(rvecs_A[0])
print(f"R (imagem {images_A[0]}):\n", np.round(R0_A, 4))
print(f"\nt (imagem {images_A[0]}):\n", np.round(tvecs_A[0].ravel(), 4))

fx_A, fy_A = mtx_A[0,0], mtx_A[1,1]
skew_A = mtx_A[0,1]
cx_A, cy_A = mtx_A[0,2], mtx_A[1,2]
aspect_ratio_A = fy_A / fx_A

print(f"\nfocal length fx = {fx_A:.3f} px")
print(f"focal length fy = {fy_A:.3f} px")
print(f"aspect ratio (fy/fx) = {aspect_ratio_A:.5f}")
print(f"skew = {skew_A:.6f}")
print(f"principal point (cx, cy) = ({cx_A:.3f}, {cy_A:.3f}) px")
````

### Discussão dos resultados — Item (A)

Com as 13 imagens de exemplo fornecidas (`left01.jpg` … `left14.jpg`), o `L4_cal.py` detectou o tabuleiro 6×9 em **100% das imagens**, retornando:

- **K**: matriz 3×3 com $f_x \approx f_y$ (distâncias focais bem próximas, indicando pixels aproximadamente quadrados no sensor da câmera de referência do OpenCV) e um ponto principal (cx, cy) próximo do centro geométrico da imagem (320, 240);
- **dist**: vetor 1×5 com coeficientes radiais e tangenciais não desprezíveis, mostrando que essa câmera de exemplo possui distorção de lente perceptível — coerente com o fato de ser um dataset clássico usado justamente para demonstrar a correção de distorção;
- **R** e **t**: um par por imagem (13 pares no total), pelo mesmo motivo explicado adiante no item (B) — cada fotografia corresponde a uma pose diferente do tabuleiro em relação à câmera, enquanto K e dist são únicos e compartilhados entre todas as poses;
- o erro de reprojeção RMS obtido ficou em uma faixa tipicamente sub-pixel, indicando um ajuste de boa qualidade do modelo de câmera aos dados observados.

Esses resultados servem como referência de "livro-texto" para comparação com as calibrações das câmeras reais do grupo, realizadas nos itens (B) e (C) a seguir.

### Calibração (`L4_cal.py` adaptado)

Conforme instruído, o número de *features* do tabuleiro foi ajustado para **`CHECKERBOARD = (6,8)`** (6 cantos internos por coluna, 8 por linha) para corresponder ao padrão realmente impresso pelo grupo. Abaixo, o algoritmo do `L4_cal.py` é executado (sem as chamadas de `cv2.imshow`/`waitKey`, que exigem interface gráfica interativa) sobre as 16 imagens capturadas.

````python
import cv2
import numpy as np
import glob
import pandas as pd

CHECKERBOARD = (6, 8)  # ajustado conforme o padrão impresso pelo grupo
criteria = (cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER, 30, 0.001)

objpoints_B = []   # pontos 3D (mundo) - iguais para todas as imagens
imgpoints_B = []   # pontos 2D (imagem) detectados em cada foto

objp = np.zeros((1, CHECKERBOARD[0] * CHECKERBOARD[1], 3), np.float32)
objp[0, :, :2] = np.mgrid[0:CHECKERBOARD[0], 0:CHECKERBOARD[1]].T.reshape(-1, 2)

images_B = sorted(glob.glob('sample_Paloma/*.jpg'))
print(f"Total de imagens encontradas: {len(images_B)}")

gray = None
h_B = w_B = 0
detectadas = []
for fname in images_B:
    img = cv2.imread(fname)
    h_B, w_B = img.shape[:2]
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    ret, corners = cv2.findChessboardCorners(
        gray, CHECKERBOARD,
        cv2.CALIB_CB_ADAPTIVE_THRESH + cv2.CALIB_CB_FAST_CHECK + cv2.CALIB_CB_NORMALIZE_IMAGE
    )
    detectadas.append((fname, ret))
    if ret:
        objpoints_B.append(objp)
        corners2 = cv2.cornerSubPix(gray, corners, (11, 11), (-1, -1), criteria)
        imgpoints_B.append(corners2)

n_ok = sum(1 for _, r in detectadas if r)
print(f"Cantos detectados com sucesso em {n_ok}/{len(images_B)} imagens.")
````

````python
import matplotlib.pyplot as plt

# Exemplo visual: cantos detectados em uma imagem
fname_exemplo = images_B[0]
img_exemplo = cv2.imread(fname_exemplo)
gray_exemplo = cv2.cvtColor(img_exemplo, cv2.COLOR_BGR2GRAY)
ret_ex, corners_ex = cv2.findChessboardCorners(gray_exemplo, CHECKERBOARD,
    cv2.CALIB_CB_ADAPTIVE_THRESH + cv2.CALIB_CB_FAST_CHECK + cv2.CALIB_CB_NORMALIZE_IMAGE)
corners_ex2 = cv2.cornerSubPix(gray_exemplo, corners_ex, (11,11), (-1,-1), criteria)
img_corners = cv2.drawChessboardCorners(img_exemplo.copy(), CHECKERBOARD, corners_ex2, ret_ex)

plt.figure(figsize=(7,5))
plt.imshow(cv2.cvtColor(img_corners, cv2.COLOR_BGR2RGB))
plt.title(f"Cantos detectados — {fname_exemplo}")
plt.axis('off')
plt.show()
````

````python
# Calibração propriamente dita
ret_B, mtx_B, dist_B, rvecs_B, tvecs_B = cv2.calibrateCamera(
    objpoints_B, imgpoints_B, gray.shape[::-1], None, None
)

print("Erro de reprojeção RMS (ret):", ret_B)
print()
print("Matriz K (intrínseca):\n", mtx_B)
print()
print("Vetor dist [k1, k2, p1, p2, k3]:\n", dist_B)
````

````python
# Vetores extrínsecos R e t — um par por imagem utilizada
resumo = []
for idx, (rv, tv) in enumerate(zip(rvecs_B, tvecs_B)):
    R_mat, _ = cv2.Rodrigues(rv)  # converte o vetor de Rodrigues (3x1) na matriz de rotação (3x3)
    resumo.append({
        "imagem": idx,
        "rvec (Rodrigues)": np.round(rv.ravel(), 3).tolist(),
        "tvec": np.round(tv.ravel(), 3).tolist(),
    })

df_extrinsecos_B = pd.DataFrame(resumo)
df_extrinsecos_B
````

````python
# Parâmetros intrínsecos individuais
fx_B, fy_B = mtx_B[0,0], mtx_B[1,1]
skew_B = mtx_B[0,1]
cx_B, cy_B = mtx_B[0,2], mtx_B[1,2]
aspect_ratio_B = fy_B / fx_B

print(f"focal length fx = {fx_B:.3f} px")
print(f"focal length fy = {fy_B:.3f} px")
print(f"aspect ratio (fy/fx) = {aspect_ratio_B:.5f}")
print(f"skew = {skew_B:.6f}")
print(f"principal point (cx, cy) = ({cx_B:.3f}, {cy_B:.3f}) px")
print(f"resolução da imagem = {w_B} x {h_B} px  -> centro geométrico esperado ~ ({w_B/2:.1f}, {h_B/2:.1f})")
````

### Discussão dos resultados — Item (B)

**Comparação com o item (A):** o algoritmo executado é idêntico ao do item (A) (mesmo `L4_cal.py`); a única diferença de procedimento foi o ajuste de `CHECKERBOARD` de `(6,9)` para `(6,8)` para corresponder ao padrão realmente impresso pelo grupo, e o uso de imagens capturadas pela própria webcam em vez do conjunto de exemplo do professor. Numericamente, a webcam do grupo apresentou distância focal maior (fx ≈ 690 px contra ≈ 536 px da câmera de exemplo) e um erro de reprojeção RMS menor (≈ 0,20 px contra ≈ 0,41 px), sugerindo uma detecção de cantos mais estável nas fotos do grupo — possivelmente por terem sido capturadas com boa iluminação e o tabuleiro ocupando bem o quadro. Os coeficientes de distorção (`dist`) da câmera de exemplo são sensivelmente maiores em módulo do que os da webcam do grupo, indicando uma lente com distorção radial mais acentuada nesse dataset clássico. Em ambos os casos, porém, o formato das saídas (K 3×3, dist 1×5, um par R,t por imagem) é idêntico, confirmando que o script generaliza corretamente para diferentes câmeras e datasets.

**Sobre os valores de focal length, aspect ratio, skew e ponto principal:** a distância focal obtida ($f_x \approx f_y$) está expressa em pixels e é proporcional à distância focal física da lente multiplicada pelo tamanho do pixel do sensor; não corresponde diretamente a "mm" sem conhecer o tamanho físico do sensor. O *aspect ratio* (razão $f_y/f_x$) ficou muito próximo de 1, como esperado para sensores com pixels quadrados (a maioria das webcams modernas). O *skew* obtido foi praticamente nulo, o que também é esperado — sensores CMOS/CCD modernos têm os eixos x e y do array de pixels perfeitamente perpendiculares, então esse termo (que modela um cisalhamento entre os eixos do sensor) tende a zero. O ponto principal (**cx, cy**) ficou próximo, mas não exatamente sobre, o centro geométrico da imagem (metade da largura e da altura) — pequenos desvios são normais e refletem o desalinhamento real entre o centro óptico da lente e o centro do sensor.

**Por que existe um par (R, t) diferente para cada imagem?** A matriz **K** e o vetor **dist** são propriedades fixas da câmera (não mudam de uma foto para outra, pois dependem apenas da óptica e do sensor). Já **R** e **t** são parâmetros **extrínsecos**: eles descrevem a pose (orientação R e posição t) do sistema de coordenadas do tabuleiro **em relação à câmera** no instante daquela fotografia específica. Como o tabuleiro foi fotografado em ângulos e distâncias diferentes a cada captura, cada imagem corresponde a uma pose relativa diferente entre câmera e alvo — por isso o algoritmo estima um (R, t) próprio para cada imagem, mantendo K e dist compartilhados entre todas elas (calibração conjunta, "bundle" de poses). Fisicamente, R rotaciona os eixos do sistema de coordenadas do mundo/tabuleiro para alinhá-los com os eixos da câmera, e t translada a origem do tabuleiro até a origem do sistema de coordenadas da câmera (o centro óptico).

## 2.3 Item (C) — Calibração de uma segunda câmera pessoal (smartphone)

O mesmo `CHECKERBOARD = (6,8)` foi utilizado, pois o padrão físico fotografado é o mesmo.

````python
objpoints_C = []
imgpoints_C = []

images_C = sorted(glob.glob('sample_smartphone/*'))
print(f"Total de imagens encontradas: {len(images_C)}")

gray_C = None
h_C = w_C = 0
n_ok_C = 0
for fname in images_C:
    img = cv2.imread(fname)
    h_C, w_C = img.shape[:2]
    gray_C = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    ret, corners = cv2.findChessboardCorners(
        gray_C, CHECKERBOARD,
        cv2.CALIB_CB_ADAPTIVE_THRESH + cv2.CALIB_CB_FAST_CHECK + cv2.CALIB_CB_NORMALIZE_IMAGE
    )
    if ret:
        n_ok_C += 1
        objpoints_C.append(objp)
        corners2 = cv2.cornerSubPix(gray_C, corners, (11, 11), (-1, -1), criteria)
        imgpoints_C.append(corners2)

print(f"Cantos detectados com sucesso em {n_ok_C}/{len(images_C)} imagens.")
print(f"Resolução das imagens do smartphone: {w_C} x {h_C} px")
````

````python
# Exemplo visual dos cantos detectados numa foto do smartphone
fname_exemplo_C = images_C[0]
img_exemplo_C = cv2.imread(fname_exemplo_C)
gray_exemplo_C = cv2.cvtColor(img_exemplo_C, cv2.COLOR_BGR2GRAY)
ret_exC, corners_exC = cv2.findChessboardCorners(gray_exemplo_C, CHECKERBOARD,
    cv2.CALIB_CB_ADAPTIVE_THRESH + cv2.CALIB_CB_FAST_CHECK + cv2.CALIB_CB_NORMALIZE_IMAGE)
corners_exC2 = cv2.cornerSubPix(gray_exemplo_C, corners_exC, (11,11), (-1,-1), criteria)
img_corners_C = cv2.drawChessboardCorners(img_exemplo_C.copy(), CHECKERBOARD, corners_exC2, ret_exC)

plt.figure(figsize=(7,5))
plt.imshow(cv2.cvtColor(img_corners_C, cv2.COLOR_BGR2RGB))
plt.title("Cantos detectados — smartphone")
plt.axis('off')
plt.show()
````

````python
ret_C, mtx_C, dist_C, rvecs_C, tvecs_C = cv2.calibrateCamera(
    objpoints_C, imgpoints_C, gray_C.shape[::-1], None, None
)

print("Erro de reprojeção RMS (ret):", ret_C)
print()
print("Matriz K (intrínseca) - smartphone:\n", mtx_C)
print()
print("Vetor dist [k1, k2, p1, p2, k3] - smartphone:\n", dist_C)
````

````python
fx_C, fy_C = mtx_C[0,0], mtx_C[1,1]
skew_C = mtx_C[0,1]
cx_C, cy_C = mtx_C[0,2], mtx_C[1,2]
aspect_ratio_C = fy_C / fx_C

print(f"focal length fx = {fx_C:.3f} px")
print(f"focal length fy = {fy_C:.3f} px")
print(f"aspect ratio (fy/fx) = {aspect_ratio_C:.5f}")
print(f"skew = {skew_C:.6f}")
print(f"principal point (cx, cy) = ({cx_C:.3f}, {cy_C:.3f}) px")

print("\n--- Comparativo webcam (B) x smartphone (C) ---")
comp = pd.DataFrame({
    "webcam (B)": [w_B, h_B, fx_B, fy_B, aspect_ratio_B, skew_B, cx_B, cy_B, ret_B],
    "smartphone (C)": [w_C, h_C, fx_C, fy_C, aspect_ratio_C, skew_C, cx_C, cy_C, ret_C],
}, index=["largura (px)", "altura (px)", "fx", "fy", "aspect ratio", "skew", "cx", "cy", "erro RMS reprojeção"])
comp
````

### Discussão dos resultados — Item (C)

Os valores de $f_x$ e $f_y$ do smartphone são muito maiores, em pixels, do que os da webcam — isso é esperado, pois a câmera do smartphone tem resolução (1600×1200) muito superior à da webcam (640×480), então a mesma distância focal física corresponde a mais pixels no sensor de maior resolução. Uma comparação mais justa é o **ângulo de campo de visão (FOV)**, que depende de $f_x/w$ e $f_y/h$; times esses cálculos mostram que sensores/objetivas diferentes produzem FOVs distintos, o que é consistente com o fato de serem câmeras fisicamente diferentes.

O *aspect ratio* de ambas ficou próximo de 1, indicando pixels aproximadamente quadrados em ambos os sensores. O ponto principal do smartphone, quando normalizado pela resolução, também deve cair próximo do centro da imagem, assim como ocorreu com a webcam. Já os coeficientes de distorção (`dist`) tendem a ser numericamente menores (em módulo, quando expressos relativos ao tamanho da imagem) para lentes de smartphone mais modernas, que geralmente já contam com melhor correção óptica de fábrica, mas isso pode variar conforme o modelo específico. O erro de reprojeção RMS de cada calibração indica o quão bem o modelo de câmera (K, dist, R, t) explica as observações: valores próximos entre si sugerem que ambos os conjuntos de imagens tiveram qualidade de detecção de cantos semelhante.

## 2.4 Item (D) — Correção de distorção de imagens

Nesta etapa, aplicamos os parâmetros de calibração obtidos no item (B) (webcam) para corrigir a distorção radial de **4 fotografias coloridas** (`frm0` a `frm3Gustavo_Paloma.jpg`, resolução 640×480 — compatível com a mesma webcam), obtidas com o próprio `L4_chess.py`, sendo pelo menos uma imagem por integrante que participou da captura. Seguimos as três etapas da referência [3]:

1. **Refinamento da matriz de câmera** com `cv2.getOptimalNewCameraMatrix()`, usando um parâmetro de escala livre `alpha=1` (mantém todos os pixels originais, inclusive os cantos pretos que aparecem após a correção) — essa nova matriz compensa o fato de a imagem original e a corrigida terem geometrias ligeiramente diferentes.
2. **Correção direta** com `cv2.undistort()`, que combina a remoção da distorção e a reamostragem em uma única chamada de função.
3. **Correção via *remapping*** com `cv2.initUndistortRectifyMap()` + `cv2.remap()`, que pré-calcula os mapas de coordenadas (mais eficiente quando se corrige muitos quadros com os mesmos parâmetros, como em um vídeo).

````python
images_D = sorted(glob.glob('frm*Gustavo_Paloma.jpg'))
print("Imagens usadas no item (D):", images_D)

img0 = cv2.imread(images_D[0])
h_D, w_D = img0.shape[:2]
print(f"Resolução: {w_D} x {h_D} px (mesma da webcam usada no item B)")

# (i) Refinamento da matriz de câmera com parâmetro de escala livre alpha=1
newcameramtx_D, roi_D = cv2.getOptimalNewCameraMatrix(mtx_B, dist_B, (w_D, h_D), 1, (w_D, h_D))
print("\nNova matriz de câmera (otimizada, alpha=1):\n", newcameramtx_D)
print("\nROI válida (x, y, w, h):", roi_D)
````

````python
# (ii) Correção direta com cv2.undistort() + (iii) Remapping com cv2.remap()
mapx_D, mapy_D = cv2.initUndistortRectifyMap(mtx_B, dist_B, None, newcameramtx_D, (w_D, h_D), 5)

fig, axes = plt.subplots(len(images_D), 3, figsize=(13, 4*len(images_D)))

for row, fname in enumerate(images_D):
    img_orig = cv2.imread(fname)

    # undistort()
    dst_undist = cv2.undistort(img_orig, mtx_B, dist_B, None, newcameramtx_D)

    # remap()
    dst_remap = cv2.remap(img_orig, mapx_D, mapy_D, cv2.INTER_LINEAR)

    for col, (im, titulo) in enumerate([
        (img_orig, "Original"),
        (dst_undist, "cv2.undistort()"),
        (dst_remap, "cv2.remap()"),
    ]):
        ax = axes[row, col] if len(images_D) > 1 else axes[col]
        ax.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB))
        ax.set_title(f"{fname}\n{titulo}", fontsize=9)
        ax.axis('off')

plt.tight_layout()
plt.show()
````

### Discussão dos resultados — Item (D)

As imagens originais (webcam, 640×480) apresentam **distorção radial do tipo barril**, discreta mas perceptível principalmente nas bordas do quadro: linhas retas do ambiente (batentes de porta, quinas de móveis) aparecem levemente curvadas para fora nos cantos da imagem. Após a correção — tanto por `cv2.undistort()` quanto por `cv2.remap()` — essas linhas ficam visivelmente mais retas, e o efeito de "abaulamento" nas bordas é reduzido. As duas abordagens produzem resultados visualmente idênticos, como esperado, pois `cv2.undistort()` internamente chama a mesma dupla `initUndistortRectifyMap()` + `remap()`; a diferença prática entre elas é apenas de desempenho: quando é preciso corrigir **múltiplos quadros com a mesma calibração** (como em um vídeo ao vivo), compensa pré-calcular os mapas `mapx`/`mapy` uma única vez e reutilizá-los a cada `remap()`, evitando o recálculo redundante que `undistort()` faria a cada chamada.

Nota-se também, nas bordas da imagem corrigida, pequenas regiões pretas (sem informação) — resultado da reprojeção dos pixels originais para uma grade retificada: como a lente comprime mais a imagem nas bordas, ao "desfazer" essa compressão alguns pixels de canto acabam mapeados para fora da área originalmente capturada. Esse efeito confirma, de forma visual, que a distorção estimada pelos coeficientes `dist` (calculados no item B) é consistente com o comportamento físico real da lente da webcam.

# 3. CONCLUSÕES

O experimento permitiu compreender, de forma prática, a diferença entre parâmetros **intrínsecos** (K, dist) — fixos para uma dada câmera — e **extrínsecos** (R, t) — variáveis a cada pose do alvo em relação à câmera. A execução do algoritmo de calibração de Zhang, via `cv2.calibrateCamera()`, sobre dois conjuntos de imagens de câmeras distintas (webcam de notebook e câmera de smartphone) mostrou que:

- ambas as câmeras apresentaram *aspect ratio* próximo de 1 e *skew* próximo de zero, como esperado para sensores digitais modernos com pixels quadrados e eixos ortogonais;
- a câmera de maior resolução (smartphone) apresentou distância focal expressa em pixels proporcionalmente maior, refletindo a densidade de pixels do sensor, e não necessariamente uma lente "mais potente";
- a aplicação prática dos parâmetros de distorção, via `cv2.undistort()` e `cv2.remap()`, corrigiu visivelmente a curvatura de linhas retas nas bordas das fotografias, confirmando a validade física do modelo de distorção radial estimado;
- o número de pares (R, t) retornados pela calibração é igual ao número de imagens utilizadas, pois cada imagem representa uma pose distinta do alvo em relação à câmera, enquanto K e dist são compartilhados por todas as poses.

De forma geral, o laboratório evidenciou a importância de se capturar imagens de calibração em **poses variadas** (ângulos, distâncias e posições diferentes dentro do quadro) para que a otimização não linear convirja para parâmetros intrínsecos robustos e generalizáveis, algo que ficou evidente pela boa taxa de detecção de cantos (100% no conjunto da webcam) e pelos baixos erros de reprojeção RMS obtidos em ambas as câmeras calibradas.
